Step 1: Rename the file to movie_data, count the review total, print column names and the row

```python
import csv

with open("IMDB Dataset.csv", "r", encoding="utf-8") as file:
    reader = csv.DictReader(file)
    movie_data = list(reader)

print("Total reviews:", len(movie_data))
print("Column names:", movie_data[0].keys())
print("First review:", movie_data[0]["review"][:300])
print("Sentiment:", movie_data[0]["sentiment"])
```

Step 2: Count the number of postive and negative reviews, then find the **Prior Probability** of negative

```python
positive_count = 0
negative_count = 0

for review in movie_data:
    if review["sentiment"] == "positive":
        positive_count += 1
    elif review["sentiment"] == "negative":
        negative_count += 1

print("Positive reviews:", positive_count)
print("Negative reviews:", negative_count)

total_reviews = len(movie_data)
prior_negative = negative_count / total_reviews
print("Prior probability of negative sentiment:", prior_negative)
```

Step 3: Pick keywords, then count the number of times a keyword appears in a review once

```python
print("Number of keywords in negative reviews:\n")
keywords = ["horrible", "disappointing", "pathetic", "boring"]
for keyword in keywords:
    negative_matches = 0

    for review in movie_data:
        if review["sentiment"] == "negative":
            if keyword in review["review"].lower():
                negative_matches +=1

    print(keyword, ":", negative_matches)
```

Step 4: Calculate the **Likehood Probalibility** of each keyword.

i.e, how likely a chosen keyword would appear in a negative review

```python
keywords = ["horrible", "disappointing", "pathetic", "boring"]

negative_keyword_counts = {
    "horrible": 1779,
    "disappointing": 639,
    "pathetic": 1106,
    "boring": 2498
}
print("Likelyhood Probability:\n")
for keyword in keywords:
    likelihood = negative_keyword_counts[keyword] / negative_count
    print(keyword,":", likelihood)
```

Step 5: Calculate probalility of a keyword appearing in either a positive or negative review. **Marginal Probability**

```python
# First, count how many of the 50,000 reviews contain each keyword.
print("Total keyword count across all reviews:")
keyword_counts = {}
for keyword in keywords:
    count = 0

    for review in movie_data:
        if keyword in review["review"].lower():
            count += 1

    keyword_counts[keyword] = count
    print(keyword,":",count)
print()
# Now, total number of reviews containing each keywords gets divided by 50,000
print("Marginal Probability:")
marginal_probabilities = {}
for keyword in keywords:
    probability = keyword_counts[keyword] / total_reviews
    marginal_probabilities[keyword] = probability

    print(keyword, ":", probability)
```

Step 6: Apply Bayes' theorem to calculate the posterior probability of each keyword. **Posterior Probability**

```python
print("Posterior Probability:\n")
for keyword in keywords:
    likelihood = negative_keyword_counts[keyword] / negative_count
    prior = prior_negative
    marginal = marginal_probabilities[keyword]

    posterior = (likelihood * prior) / marginal
    print(keyword, ":", round(posterior, 4))
```
